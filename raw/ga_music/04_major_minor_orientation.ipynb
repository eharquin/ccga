# 04 — Major/Minor Orientation

**Hypothesis:** Reversing the grade-3 trivector of a major triad yields its negative-harmony minor partner. Specifically, `~(C ∧ E ∧ G)` should equal (up to sign/scalar) the F-minor triad — the Levy polar minor of C-major.

```python
from kingdon import Algebra
import numpy as np

alg = Algebra(7, 5, 1)
def mv(k): return alg.multivector({k: 1})

PITCH = {
    'C':  mv('e1'), 'D':  mv('e2'), 'E':  mv('e3'), 'F':  mv('e4'),
    'G':  mv('e5'), 'A':  mv('e6'), 'B':  mv('e7'),
    'Cs': mv('e8'), 'Ds': mv('e9'), 'Fs': mv('eA'), 'Gs': mv('eB'), 'As': mv('eC'),
}
OCTAVE = mv('e0')

def norm2(x): return (x * ~x).e
def is_zero(x, tol=1e-10): return all(abs(float(v)) <= tol for v in x.values())
def blade_coeffs(x, tol=1e-10): return {alg.bin2canon[k]: float(v) for k, v in x.items() if abs(float(v)) > tol}

print('Setup ready.')
```

```python
# C-major trivector and its reverse
C_major_tv = PITCH['C'] ^ PITCH['E'] ^ PITCH['G']
rev_C_major = ~C_major_tv

print('C_major:', blade_coeffs(C_major_tv))
print('~C_major:', blade_coeffs(rev_C_major))
print()
# For a grade-3 blade: ~B = (-1)^(3*2/2) * B = -B
print('~(grade-3 blade) = -blade: ✓' if blade_coeffs(rev_C_major) == {k: -v for k, v in blade_coeffs(C_major_tv).items()} else '✗')
```

```python
# Levy negative harmony: polar minor of C-major around the C-G axis
# Interval inversion: C stays, E->Ab (maj3 up → maj3 down), G->F (P5 up → P5 down)
# Polar minor = F Ab C = F-minor (first inversion)
F_minor_tv = PITCH['F'] ^ PITCH['Gs'] ^ PITCH['C']

print('F_minor:', blade_coeffs(F_minor_tv))
print('~C_major:', blade_coeffs(rev_C_major))
print()

# Check if they share the same blade
same_blade = set(blade_coeffs(F_minor_tv).keys()) == set(blade_coeffs(rev_C_major).keys())
print('✓ same blade (proportional)' if same_blade else '✗ different blades — ~C_major ≠ F_minor')
```

```python
# Why they cannot be proportional: they live on different basis blades
print('C_major blade:', list(blade_coeffs(C_major_tv).keys()))
print('F_minor blade:', list(blade_coeffs(F_minor_tv).keys()))
print()
print('C_major: e1(C) ∧ e3(E) ∧ e5(G)  — all positive subspace')
print('F_minor: e4(F) ∧ eB(Gs) ∧ e1(C) — mixes positive and negative subspace')
print()
print('No inner automorphism can map one to the other:')
print('rotors preserve subspace membership, boosts cannot map grade-3→grade-3 across subspaces.')
```

```python
# Explore other operations: dual and conjugate
dual_C_major = C_major_tv.dual()
conj_C_major = C_major_tv.conjugate()
invol_C_major = C_major_tv.involute()

print('dual(C_major):', blade_coeffs(dual_C_major))
print('conjugate(C_major):', blade_coeffs(conj_C_major))
print('involute(C_major):', blade_coeffs(invol_C_major))
print()
# dual maps grade-3 → grade-10 in a 13D algebra; that cannot equal F_minor (grade-3)
dual_blade_name = list(blade_coeffs(dual_C_major).keys())[0] if blade_coeffs(dual_C_major) else 'e'
dual_grade = len(dual_blade_name.replace('e', ''))
print(f'dual(C_major) blade: {dual_blade_name}  → grade {dual_grade} — cannot equal F_minor (grade 3)')
```

```python
# Norm2 comparison
print(f'‖C_major‖²  = {norm2(C_major_tv)}')
print(f'‖C_minor‖²  = {norm2(PITCH["C"] ^ PITCH["Ds"] ^ PITCH["G"])}')
print(f'‖F_minor‖²  = {norm2(F_minor_tv)}')
print()
print('C_major (+1) and F_minor (−1) have different norms.')
print('No scalar multiple of C_major can equal F_minor — they have opposite norm² signs.')
```

## Discussion

**Failed** — informative null result.

`~(C ∧ E ∧ G)` is simply `-(C ∧ E ∧ G)` — a sign flip on the same blade `e135`. The F-minor triad `F ∧ G♯ ∧ C` occupies the blade `e14B1` (involving eB = G♯, a negative-subspace element). These are fundamentally different basis blades.

Moreover, `‖C_major‖² = +1` and `‖F_minor‖² = −1` (F_minor has one black key G♯). No scalar multiple can reconcile this sign difference.

The negative-harmony relation (Levy's polar minor) is a pitch-class symmetry on the chromatic circle — it requires mapping white keys to black keys and vice versa. This necessarily crosses the positive/negative subspace boundary of Cl(7,5,1). No inner automorphism (rotor sandwich, reverse, conjugate, dual) can achieve this crossing while preserving the grade and type of the chord.

**This is not a design flaw.** The algebra was built to distinguish tonal stability, not to represent chromatic symmetry. Capturing negative harmony would require a different algebraic structure.
