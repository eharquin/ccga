# 02 — Octave Equivalence

**Hypothesis:** The null vector `OCTAVE` (e0, e² = 0) acts as an ideal element analogous to PGA's point at infinity: adding multiples of it to a chord voicing does not change the chord's squared norm.

```python
from kingdon import Algebra
import numpy as np

alg = Algebra(7, 5, 1)
def mv(k): return alg.multivector({k: 1})

PITCH = {
    'C':  mv('e1'), 'D':  mv('e2'), 'E':  mv('e3'), 'F':  mv('e4'),
    'G':  mv('e5'), 'A':  mv('e6'), 'B':  mv('e7'),
    'Cs': mv('e8'), 'Ds': mv('e9'), 'Fs': mv('eA'), 'Gs': mv('eB'), 'As': mv('eC'),
}
OCTAVE = mv('e0')

def norm2(x): return (x * ~x).e
def is_zero(x, tol=1e-10): return all(abs(float(v)) <= tol for v in x.values())
def blade_coeffs(x, tol=1e-10): return {alg.bin2canon[k]: float(v) for k, v in x.items() if abs(float(v)) > tol}

print('Setup ready. e0 is null (OCTAVE).')
```

```python
# Claim: OCTAVE is null
oct_sq = (OCTAVE * OCTAVE).e
print(f'OCTAVE² = {oct_sq}')
print('✓ null' if oct_sq == 0 else '✗ not null')
```

```python
# Build C-major in root position and octave-shifted voicings
C, E, G = PITCH['C'], PITCH['E'], PITCH['G']

C_oct = C + OCTAVE
E_oct = E + OCTAVE
G_oct = G + OCTAVE

chord_root = C ^ E ^ G
chord_v1   = C ^ E ^ G_oct
chord_v2   = C ^ E_oct ^ G_oct
chord_v3   = C_oct ^ E_oct ^ G_oct

print('Voicings constructed.')
print('Root:', blade_coeffs(chord_root))
print('V1:  ', blade_coeffs(chord_v1))
```

```python
# Claim: squared norms are equal across voicings
norms = {'root': norm2(chord_root), 'v1': norm2(chord_v1),
         'v2': norm2(chord_v2), 'v3': norm2(chord_v3)}
for name, n in norms.items():
    print(f'  ‖{name}‖² = {n}')
print('✓ all norms equal' if len(set(norms.values())) == 1 else '✗ norms differ')
```

```python
# Claim: voicings differ only in blade components that involve e0
diff = chord_v1 - chord_root
diff_bc = blade_coeffs(diff)
print('V1 − root non-zero components:')
print(diff_bc)
only_oct = all('0' in k for k in diff_bc.keys())
print('✓ difference involves only e0 terms' if only_oct else '✗ difference has non-octave terms')
```

```python
# OCTAVE ^ OCTAVE = 0 (null element squares to zero)
oct_wedge = OCTAVE ^ OCTAVE
print(f'OCTAVE ^ OCTAVE = {blade_coeffs(oct_wedge) or 0}')
print('✓ null wedge = 0' if is_zero(oct_wedge) else '✗')
```

```python
# Claim: double octave shift also preserves norm
C_oct2 = C + 2*OCTAVE
chord_v4 = C_oct2 ^ E ^ G
n2_v4 = norm2(chord_v4)
print(f'‖chord with 2·OCTAVE on C‖² = {n2_v4}')
print('✓ same norm' if n2_v4 == norms['root'] else '✗ norm changed')
```

```python
# PGA analogy: inner product of any pitch with OCTAVE
inner_C    = (C | OCTAVE).e
inner_Coct = (C_oct | OCTAVE).e
print(f'C       | OCTAVE = {inner_C}')
print(f'C+OCTAVE | OCTAVE = {inner_Coct}')
print('(both 0 — null direction is metrically invisible)')
```

## Discussion

The null vector `OCTAVE` (e0² = 0) is metrically invisible: any blade component involving e0 contributes 0 to the squared norm. Octave-shifted voicings differ only in their e0-containing blade components — the "ideal" part of the multivector — while the metric-relevant content is unchanged.

This mirrors PGA exactly: ideal points (at infinity, i.e., carrying the null direction) do not contribute to metric quantities. The analogy is clean — the null generator is the "octave direction" in the algebra.

**Held.** Octave-shifted voicings are metrically equivalent. They are not algebraically identical (different blade coefficients), but their squared norms agree.
