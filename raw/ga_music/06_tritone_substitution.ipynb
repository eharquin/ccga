# 06 — Tritone Substitution

**Hypothesis:** In the sub-algebra spanned by the chord tones of G7 and D♭7, multiplication by the local pseudoscalar maps one chord to the other (up to sign/scalar). The regressive-product form of duality is used throughout.

```python
from kingdon import Algebra
import numpy as np

alg = Algebra(7, 5, 1)
def mv(k): return alg.multivector({k: 1})

PITCH = {
    'C':  mv('e1'), 'D':  mv('e2'), 'E':  mv('e3'), 'F':  mv('e4'),
    'G':  mv('e5'), 'A':  mv('e6'), 'B':  mv('e7'),
    'Cs': mv('e8'), 'Ds': mv('e9'), 'Fs': mv('eA'), 'Gs': mv('eB'), 'As': mv('eC'),
}
OCTAVE = mv('e0')

def norm2(x): return (x * ~x).e
def is_zero(x, tol=1e-10): return all(abs(float(v)) <= tol for v in x.values())
def blade_coeffs(x, tol=1e-10): return {alg.bin2canon[k]: float(v) for k, v in x.items() if abs(float(v)) > tol}
def regressive(A, B): return (A.dual() ^ B.dual()).dual()

print('Setup ready.')
```

```python
# G7  = G B D F   — guide tones: B(e7) and F(e4), tritone apart
# Db7 = Cs F Gs B — guide tones: F(e4) and B(e7)  (same tritone!)
G_dom7  = PITCH['G'] ^ PITCH['B']  ^ PITCH['D']  ^ PITCH['F']
Db_dom7 = PITCH['Cs'] ^ PITCH['F'] ^ PITCH['Gs'] ^ PITCH['B']

print('G7  = G B D F   :', blade_coeffs(G_dom7))
print('Db7 = Cs F Gs B :', blade_coeffs(Db_dom7))
print()
print(f'‖G7‖²  = {norm2(G_dom7)}')
print(f'‖Db7‖² = {norm2(Db_dom7)}')
```

```python
# Shared guide tones: F and B
shared = PITCH['F'] ^ PITCH['B']
print('Shared tritone F∧B:', blade_coeffs(shared))
sq_shared = norm2(shared)
print(f'‖F∧B‖² = {sq_shared}  (F white, B white → +1)')
```

```python
# Local pseudoscalar: wedge of all 6 distinct combined tones
# G7∪Db7 = {D, F, G, B, Cs, Gs} = {e2, e4, e5, e7, e8, eB}
local_ps = PITCH['D'] ^ PITCH['F'] ^ PITCH['G'] ^ PITCH['B'] ^ PITCH['Cs'] ^ PITCH['Gs']

print('Local pseudoscalar (6-blade of combined chord tones):', blade_coeffs(local_ps))
ps_sq = norm2(local_ps)
print(f'‖local_ps‖² = {ps_sq}')
print('Expected +1 (2 black keys Cs, Gs → (−1)^2 = +1):', '✓' if ps_sq == 1 else '✗')

ps_inv = ~local_ps * (1.0 / ps_sq)
```

```python
# Check that G7 ∧ Db7 spans the local pseudoscalar
# They share F(e4) and B(e7) → their outer product is ZERO (shared factors)
combined = G_dom7 ^ Db_dom7
combined_bc = blade_coeffs(combined)
local_bc    = blade_coeffs(local_ps)

print('G7 ^ Db7:', combined_bc or '0  ← zero because they share F and B')
print('local_ps:', local_bc)
print()
print('G7 and Db7 share the tritone F∧B: their wedge product vanishes.')
print('They are NOT complementary in the 6D sub-algebra — they overlap in exactly the shared guide tones.')
```

```python
# Local duality: right-multiply by ps^{-1}
# ps is invertible (ps_sq ≠ 0); ps^{-1} = ~ps / ps_sq
ps_inv = ~local_ps * (1.0 / ps_sq)

dual_G7  = G_dom7  * ps_inv
dual_Db7 = Db_dom7 * ps_inv

print('Grade note: dual of grade-4 in 6D sub-algebra → grade 6-4 = 2')
print()
print('dual_local(G7):', blade_coeffs(dual_G7))
print('dual_local(Db7):', blade_coeffs(dual_Db7))
```

## Discussion

G7 (G B D F) and D♭7 (C♯ F G♯ B) share their tritone guide tones F and B. Because they share these two elements, their outer product vanishes: `G7 ∧ D♭7 = 0`. They are *not* complementary blades — they overlap.

The 6D sub-algebra containing all 6 distinct combined tones has a well-defined pseudoscalar (norm² = +1). The local dual maps:
- `dual_local(G7) = e8B` — the Cs∧Gs blade (the two black-key tones unique to D♭7)
- `dual_local(D♭7) = e25` — the D∧G blade (the two white-key tones unique to G7)

These are neither the shared tritone F∧B nor proportional to each other. Local duality extracts the *complement* of each chord within the 6-note sub-algebra: the dual of G7 is what D♭7 contributes uniquely, and vice versa.

**Failed as stated — but informative.** Tritone substitution is not a duality in the pseudoscalar sense. What the algebra *does* show: the two chords are related by a complementarity in the 6D sub-algebra — each chord's local dual is the unique contribution of the other. The shared F∧B tritone is the intersection, not the dual. The wedge being zero is the algebraic signature of the shared guide tones that define the substitution.
