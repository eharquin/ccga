# Symbolic point pair

`pp = p1 ^ p2` with symbolic conic points carrying a radius `r` (here `r` adds
`r^2/2` to **both** `einf1` and `einf2`, i.e. `+ r^2 * einf`).

`null_coeffs_sym` / `null_coeffs_sym2` express a grade-1 / grade-2 multivector in
the null basis `eo1 eo2 eo3 e1 e2 einf1 einf2 einf3`.

Select the **CCGA (.venv)** kernel.

```python
import sympy as sp, itertools
from ccga.algebra import e1, e2, eo, eo1, eo2, eo3, einf, einf1, einf2, einf3

# reciprocal frame of the 8 null basis vectors (eo_i.einf_i = -1, e_i.e_i = 1)
_recip = {'eo1': -einf1, 'eo2': -einf2, 'eo3': -einf3, 'e1': e1, 'e2': e2,
          'einf1': -eo1, 'einf2': -eo2, 'einf3': -eo3}
_names = ['eo1', 'eo2', 'eo3', 'e1', 'e2', 'einf1', 'einf2', 'einf3']

def null_coeffs_sym(mv):
    # grade-1 null-basis coefficients
    out = {}
    for name, r in _recip.items():
        c = sp.nsimplify(sp.simplify((mv | r).e))
        if c != 0:
            out[name] = c
    return out

def null_coeffs_sym2(mv):
    # grade-2 null-bivector coefficients: coeff of n_i^n_j = mv . (recip_j ^ recip_i)
    out = {}
    for i, j in itertools.combinations(range(8), 2):
        ni, nj = _names[i], _names[j]
        c = sp.nsimplify(sp.simplify((mv | (_recip[nj] ^ _recip[ni])).e))
        if c != 0:
            out[f'{ni}^{nj}'] = c
    return out

def scalar(mv):
    return sp.nsimplify(sp.simplify(mv.e))

def sym_point(cx, cy, r):
    return (eo + cx*e1 + cy*e2
            + ((cx**2 + r**2)/2)*einf1 + ((cy**2 + r**2)/2)*einf2 + cx*cy*einf3)
```

## p1

```python
x1, y1, r1, x2, y2, r2 = sp.symbols('x1 y1 r1 x2 y2 r2', real=True)

p1 = sym_point(x1, y1, r1)
null_coeffs_sym(p1)
```

## p2

```python
p2 = sym_point(x2, y2, r2)
null_coeffs_sym(p2)
```

## pp = p1 ^ p2  (grade 2)

```python
pp = p1 ^ p2
null_coeffs_sym2(pp)            # grade-2 reader (pp is a bivector)
```

## pp^2  (reality)

```python
sp.factor(scalar(pp * pp))
```
